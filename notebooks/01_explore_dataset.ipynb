```python
# The check above only catches NaN/null — but the describe() output reveals
# suspicious minimum values (e.g. -9999) that are sentinel codes for missing data.
# Let's scan all numeric columns for any negative values.

negative_summary = {}
for col in df.select_dtypes(include='number').columns:
    n_neg = (df[col] < 0).sum()
    if n_neg > 0:
        negative_summary[col] = {'count': n_neg, 'min': df[col].min()}

pd.DataFrame(negative_summary).T.rename(columns={'count': 'negative_rows', 'min': 'min_value'})
```

Expected to be negative: latitude, longitude, temperature_celsius, temperature_fahrenheit, feels_like_celsius, and feels_like_fahrenheit can all legitimately take negative values.

Not expected to be negative: air_quality_Carbon_Monoxide, air_quality_Sulphur_dioxide, and air_quality_PM10 are physical concentrations that cannot be negative. These values (e.g. -9999) are sentinel codes representing missing data.

Whether these need replacement depends on the model: tree-based models (XGBoost, LightGBM, RF) can split on arbitrary values and handle sentinels natively, so leaving them may even help the model learn a "data-missing" signal. Linear and distance-based models (regression, SVM, k-NN) would interpret -9999 as a real magnitude, distorting coefficients and distances. For those, replacement with NaN (or imputation) is required. --> Great task for the experiments 

```python
# Let's have a look at rows where any unexpected column has a negative value
unexpected_neg_cols = ['air_quality_Carbon_Monoxide', 'air_quality_Sulphur_dioxide', 'air_quality_PM10']
mask = (df[unexpected_neg_cols] < 0).any(axis=1)

print(f"Rows with unexpected negative values: {mask.sum()} / {len(df)}\n")
df.loc[mask, ['country', 'location_name'] + unexpected_neg_cols]

# save index of these rows
suspicious = list(df.index[mask])
```

Only 4 rows out of 150k — unlikely to materially distort a linear regression. We can verify this later by fitting with and without those rows.

What *is* suspicious: **air_quality_PM10** has **two different** negative values (-1848.15 and -998.15). If these were standard sentinels we'd expect a single repeated magic number (like -9999 in the CO and SO₂ columns). Two distinct negatives in a concentration column suggests these are more likely data entry errors, instrument calibration artifacts, or formula/unit-conversion bugs in the source pipeline rather than intentional missing-data codes.

## Other extreme Outliers that seem suspicious 

```python
# Physical plausibility checks — flag values that exceed realistic bounds
outlier_checks = {
    'temperature_celsius': {'max': 60, 'reason': 'highest recorded on Earth: 56.7°C'},
    'wind_mph': {'max': 250, 'reason': 'strongest tornado winds ~300 mph'},
    'wind_kph': {'max': 400, 'reason': 'strongest tornado winds ~480 kph'},
    'gust_mph': {'max': 250, 'reason': 'same as wind'},
    'gust_kph': {'max': 400, 'reason': 'same as wind'},
    'pressure_mb': {'max': 1084, 'reason': 'highest ever recorded: 1084 mb'},
}

print("=== Values exceeding physical plausibility thresholds ===\n")
for col, bounds in outlier_checks.items():
    n = (df[col] > bounds['max']).sum()
    if n > 0:
        vals = df.loc[df[col] > bounds['max'], col].values
        print(f"  {col}: {n} rows > {bounds['max']} ({bounds['reason']})")
        print(f"    values: {sorted(vals)}\n")

# save the index of these rows to the previously made list
for col, bounds in outlier_checks.items():
    idxs = df.index[df[col] > bounds['max']].tolist()
    suspicious.extend(idxs)
```

```python
# Show ALL rows flagged as suspicious
print(f"Total suspicious entries: {len(suspicious)} (unique rows: {len(set(suspicious))})\n")
df.loc[list(set(suspicious))]
```

**Key insights for the experiments:**

This notebook serves as preparation for the MAS experiments. It is not for cleaning or modeling the data ourselves here. The goal is to understand the dataset deeply enough to (a) anticipate what agents might do and (b) later establish a ground-truth baseline to evaluate their results against.

### What could happen in the experiments?

The 8 suspicious rows present a realistic data quality challenge. Several outcomes are possible when agents encounter them:

1. **Agents ignore the issues entirely.** They skip EDA or outlier checks or ignore the outputs of their EDA and proceed straight to modeling. The impact on model performance is likely small (8/150k), but it reveals a lack of thoroughness. For the smoke test (top 5 hottest cities), ignoring data quality leads to a misleading ranking dominated by single-observation cities, a more visible failure.

2. **Agents detect and drop all suspicious rows.** This is a perfectly valid approach. Dropping 8 rows is negligible. The risk is minimal, and it shows the agents recognized something was wrong. However, a smarter approach would distinguish between sentinel codes (-9999, which are clearly missing-data markers) and physically implausible values (1841 mph wind, 79.3°C temperature), since these have different root causes, which they are expected to discuss. 

3. **Agents detect and replace/impute specific values.** The most nuanced approach: replace sentinels with NaN, cap or remove physically impossible values, and keep the rest of the row intact. This preserves valid data in other columns. Whether agents adapt their strategy to the downstream model (tree-based vs. linear) would be an additional quality signal.

4. **Agents get stuck in an endless discussion.** In a MAS setting, different agents may have conflicting opinions: one agent might argue for dropping rows, another for imputation, a third for leaving them as-is. If the system lacks a clear decision protocol, this could lead to circular debate without resolution. This is itself an interesting experimental observation about MAS coordination.

Let's establish a **ground-truth baseline**: clean the data using best practices, run the same tasks the agents will receive, and record the best achievable results. This gives a threshold to evaluate whether the MAS produced good, mediocre, or poor output, not just whether it "did something" but whether what it did was actually correct.

```python
# --- Build cleaned dataframe: df_clean ---
df_clean = df.copy()

# 1. Fix country name duplicates (based on fuzzy matching + manual verification)
country_mapping = {
    'Bélgica': 'Belgium',                              # Portuguese
    'Estonie': 'Estonia',                               # French
    'Inde': 'India',                                    # French
    'Jemen': 'Yemen',                                   # German
    'Komoren': 'Comoros',                                # German
    'Letonia': 'Latvia',                                 # Spanish/Portuguese
    'Malásia': 'Malaysia',                               # Portuguese
    'Marrocos': 'Morocco',                               # Portuguese
    'Mexique': 'Mexico',                                 # French
    'Polônia': 'Poland',                                 # Portuguese
    'Saint-Vincent-et-les-Grenadines': 'Saint Vincent and the Grenadines',  # French
    'Saudi Arabien': 'Saudi Arabia',                     # German
    'Südkorea': 'South Korea',                           # German
    'Turkménistan': 'Turkmenistan',                      # French
    'USA United States of America': 'United States of America',  # concatenation artifact
    'Гватемала': 'Guatemala',                            # Russian
    'Польша': 'Poland',                                  # Russian
    'Турция': 'Turkey',                                  # Russian
    'كولومبيا': 'Colombia',                              # Arabic
    '火鸡': 'Turkey',                                     # Chinese
}
df_clean['country'] = df_clean['country'].astype(str).str.strip().replace(country_mapping)

# 2. Fix city name duplicates (verified by checking country association per location)
city_mapping = {
    # Clear duplicates — small entry merges into main entry
    'Addis Abeba': 'Addis Ababa',                        # alternate spelling (2 → 770 rows)
    'Ar Riyadh': 'Riyadh',                               # alternate name (1 → 772 rows)
    'Beijing Shi': 'Beijing',                            # Chinese suffix 市 (1 → 770 rows)
    'City Of San Marino': 'San Marino',                  # verbose form (1 → 769 rows)
    'Kuwait': 'Kuwait City',                             # short form (1 → 771 rows)
    'Mexico (Grupo Mexico)': 'Mexico City',              # parsing artifact (1 → 769 rows)
    'New Guatemala': 'Guatemala City',                   # translation (1 → 766 rows)
    'Phnum Penh': 'Phnom Penh',                          # alternate transliteration (2 → 770 rows)
    # Fix spelling — large entries with incorrect/truncated names
    'Bras': 'Brasilia',                                  # truncated (767 rows)
    "Nuku`Aloia": "Nuku'alofa",                          # fix to correct spelling (769 rows)
    'Rangoon': 'Yangon',                                 # old colonial name → modern name (307 + 424 rows)
}
# NOTE: The following are NOT duplicates (verified via country column):
#   Bogot (Bulgaria/Hungary) ≠ Bogota (Colombia)
#   Lom (Bulgaria/Russia/Norway) ≠ Lome (Togo)
#   Laos (Indonesia) — a place in Indonesia, not the country
#   Ivory (Madagascar) — a place in Madagascar, not "Ivory Coast"
#   Chi Chi Khvar (Iran) ≠ Chi-Chi-Erh (Russia)
#   Nan (Thailand) — real city in Thailand
#   National (Bolivia) — actual location name
# NOTE: '-Kingdom' (1 row, country=火鸡, coords 38.85/34.65 = central Turkey) is a
#   corrupted entry, not a UK parsing artifact. Left unmapped (1 garbage row).
df_clean['location_name'] = df_clean['location_name'].astype(str).str.strip().replace(city_mapping)

# 3. Replace sentinel values with NaN
sentinel_cols = ['air_quality_Carbon_Monoxide', 'air_quality_Sulphur_dioxide', 'air_quality_PM10']
for col in sentinel_cols:
    df_clean.loc[df_clean[col] < 0, col] = np.nan

# 4. Remove physically implausible rows (temperature > 60°C, wind > 250 mph, pressure > 1084 mb)
implausible_mask = (
    (df_clean['temperature_celsius'] > 60) |
    (df_clean['wind_mph'] > 250) |
    (df_clean['gust_mph'] > 250) |
    (df_clean['pressure_mb'] > 1084)
)
print(f"Rows removed (physically implausible and would skew analysis): {implausible_mask.sum()}")
df_clean = df_clean[~implausible_mask].reset_index(drop=True)

# Summary
print(f"\n--- Cleaning summary ---")
print(f"Original:  {len(df)} rows, {df['country'].nunique()} countries, {df['location_name'].nunique()} cities")
print(f"Cleaned:   {len(df_clean)} rows, {df_clean['country'].nunique()} countries, {df_clean['location_name'].nunique()} cities")
print(f"\nCountry mappings applied: {len(country_mapping)} fixes")
print(f"City mappings applied: {len(city_mapping)} fixes")
print(f"Sentinel values replaced with NaN in: {sentinel_cols}")
```

## Temporal Coverage

```python
# ── Temporal Coverage: Full Exploration ──────────────────────────────────────
# Goal: Understand the temporal granularity (hourly? daily? irregular?)
#        and how coverage changes after name deduplication (duplicates at same
#        timestamp? gaps that get filled?)
# Also: plot temperature timelines to understand forecasting challenge.

# Ensure datetime column exists on both dataframes
df['last_updated_dt'] = pd.to_datetime(df['last_updated'])
df_clean['last_updated_dt'] = pd.to_datetime(df_clean['last_updated'])

# ═══════════════════════════════════════════════════════════════════════════════
# PART 1: RAW DATA (df) — what the agents receive
# ═══════════════════════════════════════════════════════════════════════════════

t_min, t_max = df['last_updated_dt'].min(), df['last_updated_dt'].max()
span = (t_max - t_min).days
print("╔══════════════════════════════════════════════════════════════╗")
print("║  RAW DATA (df) — temporal structure                          ║")
print("╚══════════════════════════════════════════════════════════════╝")
print(f"  Date range : {t_min} → {t_max}")
print(f"  Total span : {span} days (~{span/30:.1f} months)")
print(f"  Total rows : {len(df):,}")
print(f"  Unique locations: {df['location_name'].nunique()}")

# --- Per-location observation counts (raw) ---
obs_raw = df.groupby('location_name')['last_updated_dt'].agg(['count', 'min', 'max'])
obs_raw['span_days'] = (obs_raw['max'] - obs_raw['min']).dt.days
obs_raw = obs_raw.sort_values('count', ascending=False)

print(f"\n  Observations per location:")
print(f"  {obs_raw['count'].describe().to_string()}")

# --- Determine granularity: gap analysis for a well-covered city ---
example_city = obs_raw.index[0]
city_df = df[df['location_name'] == example_city].sort_values('last_updated_dt').copy()
city_df['gap_h'] = city_df['last_updated_dt'].diff().dt.total_seconds() / 3600

print(f"\n  ── Gap analysis for '{example_city}' ({len(city_df)} obs) ──")
print(f"     Median gap: {city_df['gap_h'].median():.1f} h")
print(f"     Mean gap:   {city_df['gap_h'].mean():.1f} h")
print(f"     Min gap:    {city_df['gap_h'].min():.2f} h")
print(f"     Max gap:    {city_df['gap_h'].max():.1f} h")

# --- Gap analysis across ALL locations ---
def location_gap_stats(g):
    g = g.sort_values('last_updated_dt')
    gaps_h = g['last_updated_dt'].diff().dt.total_seconds() / 3600
    return pd.Series({
        'n_obs': len(g),
        'median_gap_h': gaps_h.median(),
        'mean_gap_h': gaps_h.mean(),
        'max_gap_h': gaps_h.max(),
    })

gap_stats_raw = df.groupby('location_name').apply(location_gap_stats, include_groups=False)
gap_stats_raw = gap_stats_raw[gap_stats_raw['n_obs'] > 1]

median_of_medians = gap_stats_raw['median_gap_h'].median()
print(f"\n  ── Aggregated gap stats (all locations with >1 obs) ──")
print(f"     Locations analyzed: {len(gap_stats_raw)}")
print(f"     Median-of-median-gaps: {median_of_medians:.1f} hours")

if median_of_medians < 2:
    freq_label = "SUB-HOURLY / HOURLY"
elif median_of_medians < 25:
    freq_label = "~DAILY (one observation per day)"
elif median_of_medians < 170:
    freq_label = "~WEEKLY"
else:
    freq_label = "MONTHLY or sparser"
print(f"     → Data is approximately: {freq_label}")

# ═══════════════════════════════════════════════════════════════════════════════
# PART 2: CLEANED DATA — how does deduplication change temporal coverage?
# Key questions: Do merged locations now have DUPLICATE timestamps?
#                Do previously gappy timelines become more complete?
# ═══════════════════════════════════════════════════════════════════════════════

print("\n\n╔══════════════════════════════════════════════════════════════╗")
print("║  CLEANED DATA — temporal coverage after name dedup           ║")
print("╚══════════════════════════════════════════════════════════════╝")

# Check for duplicate timestamps within the same location (after merging)
dupes_per_loc = df_clean.groupby('location_name')['last_updated_dt'].apply(
    lambda x: x.duplicated().sum()
)
locs_with_dupes = dupes_per_loc[dupes_per_loc > 0]
print(f"\n  Locations with duplicate timestamps after dedup: {len(locs_with_dupes)}")
if len(locs_with_dupes) > 0:
    print(f"  (This means merged names had overlapping data at the same time!)")
    for loc, n in locs_with_dupes.sort_values(ascending=False).head(10).items():
        print(f"    {loc}: {n} duplicate timestamps")

# Compare gap structure: raw vs clean for locations that were affected by merging
gap_stats_clean = df_clean.groupby('location_name').apply(location_gap_stats, include_groups=False)
gap_stats_clean = gap_stats_clean[gap_stats_clean['n_obs'] > 1]

print(f"\n  ── Gap comparison: raw vs clean ──")
print(f"     Raw:   median-of-median-gaps = {gap_stats_raw['median_gap_h'].median():.1f} h")
print(f"     Clean: median-of-median-gaps = {gap_stats_clean['median_gap_h'].median():.1f} h")
print(f"\n     Raw:   mean max-gap per location = {gap_stats_raw['max_gap_h'].mean():.1f} h")
print(f"     Clean: mean max-gap per location = {gap_stats_clean['max_gap_h'].mean():.1f} h")

# Show gap stats for locations that exist in clean but had merged names
# (compare their coverage before/after)
common_locs = gap_stats_raw.index.intersection(gap_stats_clean.index)
gap_diff = gap_stats_clean.loc[common_locs, 'median_gap_h'] - gap_stats_raw.loc[common_locs, 'median_gap_h']
improved = gap_diff[gap_diff < -1]  # gap shrank by >1h = timeline got denser
worsened = gap_diff[gap_diff > 1]   # shouldn't happen, but check

print(f"\n     Locations where median gap DECREASED (>1h) after clean: {len(improved)}")
print(f"     Locations where median gap INCREASED (>1h) after clean: {len(worsened)}")
if len(improved) > 0:
    print(f"     → These had gaps filled by merging duplicate city names")
    for loc in improved.sort_values().head(5).index:
        print(f"       {loc}: {gap_stats_raw.loc[loc, 'median_gap_h']:.1f}h → {gap_stats_clean.loc[loc, 'median_gap_h']:.1f}h")
```

```python
# ── Temporal Coverage: Visualizations ────────────────────────────────────────

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# --- Plot 1: Observations per month — RAW (bar chart) ---
ax = axes[0]
df['year_month'] = df['last_updated_dt'].dt.to_period('M').astype(str)
month_counts_raw = df.groupby('year_month').size()
month_counts_raw.plot(kind='bar', ax=ax, color='salmon', edgecolor='black', alpha=0.8)
ax.set_title('Observations per month — RAW (df)', fontweight='bold')
ax.set_xlabel('')
ax.set_ylabel('Row count')
ax.tick_params(axis='x', rotation=45, labelsize=7)

# --- Plot 2: Unique locations per month — RAW vs CLEAN (line chart) ---
ax = axes[1]
locs_per_month_raw = df.groupby('year_month')['location_name'].nunique()
df_clean['year_month'] = df_clean['last_updated_dt'].dt.to_period('M').astype(str)
locs_per_month_clean = df_clean.groupby('year_month')['location_name'].nunique()
ax.plot(locs_per_month_raw.index, locs_per_month_raw.values, marker='o', color='salmon',
        lw=2, label='Raw (df)')
ax.plot(locs_per_month_clean.index, locs_per_month_clean.values, marker='s', color='steelblue',
        lw=2, label='Clean (df_clean)')
ax.set_title('Unique locations per month — RAW vs CLEAN', fontweight='bold')
ax.set_xlabel('')
ax.set_ylabel('Number of unique locations')
ax.tick_params(axis='x', rotation=45, labelsize=7)
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ── Observation timeline: well-covered vs. sparse cities (RAW) ───────────────

fig, ax = plt.subplots(figsize=(16, 8))

well_covered = list(obs_raw[obs_raw['count'] >= 700].sample(5, random_state=42).index)
sparse = list(obs_raw[(obs_raw['count'] > 1) & (obs_raw['count'] < 100)].head(5).index)
sample_cities = well_covered + sparse

colors = plt.cm.tab10(range(len(sample_cities)))

for i, city in enumerate(sample_cities):
    city_times = df.loc[df['location_name'] == city, 'last_updated_dt'].sort_values()
    ax.scatter(city_times, [i]*len(city_times), s=14, alpha=0.5, color=colors[i],
               label=f"{city} (n={len(city_times)})")

ax.set_yticks(range(len(sample_cities)))
ax.set_yticklabels(sample_cities, fontsize=9)
ax.axhline(4.5, color='gray', ls=':', lw=1.5, alpha=0.7)
ax.text(df['last_updated_dt'].max(), 2, 'well-covered', ha='right', va='center',
        fontsize=9, fontstyle='italic', color='green', alpha=0.7)
ax.text(df['last_updated_dt'].max(), 7, 'sparse', ha='right', va='center',
        fontsize=9, fontstyle='italic', color='red', alpha=0.7)
ax.set_title('Observation timeline: well-covered vs. sparse cities (raw df)', fontweight='bold', fontsize=13)
ax.set_xlabel('Date')
ax.legend(loc='upper left', fontsize=7, bbox_to_anchor=(0, -0.08), ncol=5)
ax.grid(True, axis='x', alpha=0.2)

plt.tight_layout()
plt.show()

# ── Impact of city-name cleaning on timelines ────────────────────────────────
# Each row = one merged city. Colors distinguish the original raw names that
# were combined. This shows how separate entries fill in the same timeline.

city_mapping_visual = {
    'Yangon': ['Rangoon', 'Yangon'],
    'Addis Ababa': ['Addis Abeba', 'Addis Ababa'],
    'Phnom Penh': ['Phnum Penh', 'Phnom Penh'],
    'Brasilia': ['Bras', 'Brasilia'],
    "Nuku'alofa": ["Nuku`Aloia", "Nuku'alofa"],
    'Riyadh': ['Ar Riyadh', 'Riyadh'],
    'Beijing': ['Beijing Shi', 'Beijing'],
    'Kuwait City': ['Kuwait', 'Kuwait City'],
    'Mexico City': ['Mexico (Grupo Mexico)', 'Mexico City'],
    'Guatemala City': ['New Guatemala', 'Guatemala City'],
    'San Marino': ['City Of San Marino', 'San Marino'],
}

fig, ax = plt.subplots(figsize=(16, 10))

i = 0
yticks = []
ylabels = []
for clean_name, raw_names in city_mapping_visual.items():
    # Plot each raw source in a different color on the same row
    for j, raw_name in enumerate(raw_names):
        times = df.loc[df['location_name'] == raw_name, 'last_updated_dt'].sort_values()
        n = len(times)
        color = 'salmon' if j == 0 else 'steelblue'
        marker_label = f"{raw_name} (n={n})" if i == 0 else None  # legend only once
        ax.scatter(times, [i]*len(times), s=14, alpha=0.5, color=color)
        # Annotate counts to the right of each row
        if j == 0:
            ax.annotate(f"{raw_name}: {n}", xy=(1.01, i), xycoords=('axes fraction', 'data'),
                        fontsize=7, color='salmon', va='center')
        else:
            ax.annotate(f"{raw_name}: {n}", xy=(1.01, i - 0.3), xycoords=('axes fraction', 'data'),
                        fontsize=7, color='steelblue', va='center')
    yticks.append(i)
    ylabels.append(f"{clean_name}")
    i += 1

ax.set_yticks(yticks)
ax.set_yticklabels(ylabels, fontsize=9)
ax.set_title('Impact of city-name cleaning: merged timelines (raw sources overlaid)',
             fontweight='bold', fontsize=13)
ax.set_xlabel('Date')
ax.grid(True, axis='x', alpha=0.2)

# Custom legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor='salmon', alpha=0.5, label='Source name (merged INTO target)'),
                   Patch(facecolor='steelblue', alpha=0.5, label='Target name (kept)')]
ax.legend(handles=legend_elements, fontsize=8, loc='lower right')

plt.subplots_adjust(right=0.78)
plt.tight_layout()
plt.show()

# ── Temperature timelines for selected cities ────────────────────────────────
# This helps understand the forecasting challenge: trend, seasonality, noise?

fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)
# Pick 3 cities from different climate zones with good coverage
obs_clean_counts = df_clean.groupby('location_name').size()
timeline_cities = list(obs_raw[obs_raw['count'] >= 700].sample(3, random_state=7).index)

for ax, city in zip(axes, timeline_cities):
    cdf = df_clean[df_clean['location_name'] == city].sort_values('last_updated_dt')
    ax.plot(cdf['last_updated_dt'], cdf['temperature_celsius'], lw=0.8, alpha=0.8, color='tomato')
    ax.set_ylabel('°C')
    ax.set_title(f'{city} — temperature_celsius over time ({len(cdf)} obs)', fontweight='bold', fontsize=10)
    ax.grid(True, alpha=0.3)
    # Add a 7-day rolling mean to show trend
    if len(cdf) > 7:
        rolling = cdf.set_index('last_updated_dt')['temperature_celsius'].rolling('7D').mean()
        ax.plot(rolling.index, rolling.values, lw=1.5, color='navy', alpha=0.7, label='7-day rolling mean')
        ax.legend(fontsize=8)

axes[-1].set_xlabel('Date')
plt.suptitle('Temperature timelines — understanding the forecasting signal', fontweight='bold', fontsize=12, y=1.01)
plt.tight_layout()
plt.show()

# ── Coverage tier summary ────────────────────────────────────────────────────
print("\n" + "="*65)
print("COVERAGE TIERS (raw vs clean)")
print("="*65)
tier_bins = [0, 1, 50, 200, 500, 800]
tier_labels = ['1 obs only', '2–50', '51–200', '201–500', '501+ (near-full)']

tiers_raw = pd.cut(obs_raw['count'], bins=tier_bins, labels=tier_labels).value_counts().sort_index()
tiers_clean = pd.cut(obs_clean_counts, bins=tier_bins, labels=tier_labels).value_counts().sort_index()

tier_comparison = pd.DataFrame({'raw (df)': tiers_raw, 'clean (df_clean)': tiers_clean})
print(tier_comparison.to_string())

# ── Locations with very few observations in df_clean ─────────────────────────
print("\n" + "="*65)
print("LOCATIONS WITH ≤10 OBSERVATIONS IN df_clean")
print("="*65)
sparse_locs = obs_clean_counts[obs_clean_counts <= 10].sort_values()
print(f"\n  {len(sparse_locs)} locations with ≤10 observations:\n")
for loc, n in sparse_locs.items():
    # also show the country for context
    country = df_clean.loc[df_clean['location_name'] == loc, 'country'].iloc[0]
    print(f"    {loc} ({country}): {n} obs")
```
