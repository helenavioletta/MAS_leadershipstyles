```python
# The check above only catches NaN/null — but the describe() output reveals
# suspicious minimum values (e.g. -9999) that are sentinel codes for missing data.
# Let's scan all numeric columns for any negative values.

negative_summary = {}
for col in df.select_dtypes(include='number').columns:
    n_neg = (df[col] < 0).sum()
    if n_neg > 0:
        negative_summary[col] = {'count': n_neg, 'min': df[col].min()}

pd.DataFrame(negative_summary).T.rename(columns={'count': 'negative_rows', 'min': 'min_value'})
```

Expected to be negative: latitude, longitude, temperature_celsius, temperature_fahrenheit, feels_like_celsius, and feels_like_fahrenheit can all legitimately take negative values.

Not expected to be negative: air_quality_Carbon_Monoxide, air_quality_Sulphur_dioxide, and air_quality_PM10 are physical concentrations that cannot be negative. These values (e.g. -9999) are sentinel codes representing missing data.

Whether these need replacement depends on the model: tree-based models (XGBoost, LightGBM, RF) can split on arbitrary values and handle sentinels natively, so leaving them may even help the model learn a "data-missing" signal. Linear and distance-based models (regression, SVM, k-NN) would interpret -9999 as a real magnitude, distorting coefficients and distances — for those, replacement with NaN (or imputation) is required. --> Great task for the experiments 

```python
# Let's have a look at rows where any unexpected column has a negative value
unexpected_neg_cols = ['air_quality_Carbon_Monoxide', 'air_quality_Sulphur_dioxide', 'air_quality_PM10']
mask = (df[unexpected_neg_cols] < 0).any(axis=1)

print(f"Rows with unexpected negative values: {mask.sum()} / {len(df)}\n")
df.loc[mask, ['country', 'location_name'] + unexpected_neg_cols]

# add code here to save index of these rows under suspicious (it should not be a set, but a list or dict or somethign so it is possible to have indices more than once)
```

Only 4 rows out of 150k — unlikely to materially distort a linear regression. We can verify this later by fitting with and without those rows.

What *is* suspicious: **air_quality_PM10** has **two different** negative values (-1848.15 and -998.15). If these were standard sentinels we'd expect a single repeated magic number (like -9999 in the CO and SO₂ columns). Two distinct negatives in a concentration column suggests these are more likely data entry errors, instrument calibration artifacts, or formula/unit-conversion bugs in the source pipeline rather than intentional missing-data codes.

## Long Task: "Full Exploratory Data Analysis"

**Goal:** Perform a comprehensive EDA on the Global Weather Repository:
1. **Clean the data** — handle sentinel/invalid values
2. **Engineer at least 2 new features** — e.g., heat index, air quality index category
3. **Produce 6+ visualizations**
4. **Write a ~1000-word analytical report** with methodology, key findings, and actionable recommendations

```python
# ============================================================
# LONG TASK — Step 1: Data Cleaning
# ============================================================
import warnings
warnings.filterwarnings('ignore')

df_clean = df.copy()

# 1a. Replace sentinel values (-9999 or large negatives) with NaN
sentinel_cols = [
    'air_quality_Carbon_Monoxide',
    'air_quality_Sulphur_dioxide',
    'air_quality_PM10',
]
for col in sentinel_cols:
    n_bad = (df_clean[col] < 0).sum()
    df_clean.loc[df_clean[col] < 0, col] = np.nan
    print(f"  {col}: replaced {n_bad} negative sentinel values with NaN")

# 1b. Cap extreme wind/gust values (likely instrument errors)
for col in ['wind_mph', 'wind_kph', 'gust_mph', 'gust_kph']:
    q99 = df_clean[col].quantile(0.99)
    n_outliers = (df_clean[col] > q99 * 3).sum()
    df_clean.loc[df_clean[col] > q99 * 3, col] = np.nan
    print(f"  {col}: capped {n_outliers} extreme outliers (> 3× the 99th percentile)")

# 1c. Cap extreme pressure values
q99_p = df_clean['pressure_mb'].quantile(0.99)
n_p = (df_clean['pressure_mb'] > q99_p * 1.5).sum()
df_clean.loc[df_clean['pressure_mb'] > q99_p * 1.5, 'pressure_mb'] = np.nan
print(f"  pressure_mb: capped {n_p} extreme outliers")

# 1d. Cap extreme temperature (>60°C is unrealistic)
n_temp = (df_clean['temperature_celsius'] > 60).sum()
df_clean.loc[df_clean['temperature_celsius'] > 60, 'temperature_celsius'] = np.nan
print(f"  temperature_celsius: capped {n_temp} values > 60°C")

# Summary after cleaning
missing_after = df_clean.isnull().sum()
missing_after = missing_after[missing_after > 0].sort_values(ascending=False)
print(f"\nAfter cleaning — columns with NaN values:")
for col, cnt in missing_after.items():
    print(f"  {col}: {cnt} NaN ({cnt/len(df_clean)*100:.2f}%)")
print(f"\nTotal rows: {len(df_clean):,} (unchanged — NaN used instead of dropping)")
```

```python
# ============================================================
# LONG TASK — Step 2: Feature Engineering (2 new features)
# ============================================================

# --- Feature 1: Heat Index (apparent temperature combining temp + humidity) ---
# Simplified Rothfusz regression (valid for T >= 27°C / 80°F and RH >= 40%)
# For other conditions, falls back to simple formula.

def compute_heat_index(row):
    """Compute heat index in °C from temperature (°C) and humidity (%)."""
    T = row['temperature_fahrenheit']  # Rothfusz uses °F
    RH = row['humidity']
    if pd.isna(T) or pd.isna(RH):
        return np.nan
    if T < 80:
        # Simple formula for lower temps
        HI = 0.5 * (T + 61.0 + ((T - 68.0) * 1.2) + (RH * 0.094))
        return (HI - 32) * 5 / 9  # Convert back to °C
    # Rothfusz regression
    HI = (
        -42.379
        + 2.04901523 * T
        + 10.14333127 * RH
        - 0.22475541 * T * RH
        - 0.00683783 * T**2
        - 0.05481717 * RH**2
        + 0.00122874 * T**2 * RH
        + 0.00085282 * T * RH**2
        - 0.00000199 * T**2 * RH**2
    )
    return (HI - 32) * 5 / 9  # Convert to °C

df_clean['heat_index_celsius'] = df_clean.apply(compute_heat_index, axis=1)
print("Feature 1: heat_index_celsius")
print(f"  Non-null: {df_clean['heat_index_celsius'].notna().sum():,}")
print(f"  Range: {df_clean['heat_index_celsius'].min():.1f}°C to {df_clean['heat_index_celsius'].max():.1f}°C")
print(f"  Mean: {df_clean['heat_index_celsius'].mean():.1f}°C")

# --- Feature 2: Air Quality Category based on US EPA index ---
# EPA index: 1=Good, 2=Moderate, 3=Unhealthy-Sensitive, 4=Unhealthy, 5=Very Unhealthy, 6=Hazardous
epa_labels = {1: 'Good', 2: 'Moderate', 3: 'Unhealthy (Sensitive)', 4: 'Unhealthy', 5: 'Very Unhealthy', 6: 'Hazardous'}
df_clean['aq_category'] = df_clean['air_quality_us-epa-index'].map(epa_labels)

print("\nFeature 2: aq_category (from US EPA index)")
print(df_clean['aq_category'].value_counts().sort_index())
print(f"\n  Clean dataset shape: {df_clean.shape}")
```
